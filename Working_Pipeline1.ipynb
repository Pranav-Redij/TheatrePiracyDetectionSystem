```python
import cv2
import numpy as np

INPUT_VIDEO = "/content/4ktest.mp4"
OUTPUT_VIDEO = "encoded_123456.mp4"

# CHANGE ONLY THIS MESSAGE
MESSAGE = [1,2,3,4,5,6]   # example: 123456

DIGIT_TO_SYMBOL = {
    1: "+",
    2: "-",
    3: "x",
    4: "|",
    5: "--",
    6: "||"
}

BLUE = (255, 0, 0)
ROI = 200
THICK = 18

INTERVAL_SEC = 5
ON_SEC = 1

def get_pos(h, w, i):
    pos = [
        (h//4, w//4),
        (h//4, 3*w//4),
        (3*h//4, 3*w//4),
        (3*h//4, w//4),
        (h//2, w//2)
    ]
    return pos[i % len(pos)]

def draw_symbol(sym):
    img = np.zeros((ROI, ROI, 3), dtype=np.uint8)
    c = ROI // 2
    m = 30

    if sym == "|":
        cv2.line(img,(c,m),(c,ROI-m),BLUE,THICK)
    elif sym == "||":
        cv2.line(img,(c-25,m),(c-25,ROI-m),BLUE,THICK)
        cv2.line(img,(c+25,m),(c+25,ROI-m),BLUE,THICK)
    elif sym == "-":
        cv2.line(img,(m,c),(ROI-m,c),BLUE,THICK)
    elif sym == "--":
        cv2.line(img,(m,c-20),(ROI-m,c-20),BLUE,THICK)
        cv2.line(img,(m,c+20),(ROI-m,c+20),BLUE,THICK)
    elif sym == "+":
        cv2.line(img,(c,m),(c,ROI-m),BLUE,THICK)
        cv2.line(img,(m,c),(ROI-m,c),BLUE,THICK)
    elif sym == "x":
        cv2.line(img,(m,m),(ROI-m,ROI-m),BLUE,THICK)
        cv2.line(img,(ROI-m,m),(m,ROI-m),BLUE,THICK)

    return img

cap = cv2.VideoCapture(INPUT_VIDEO)
fps = cap.get(cv2.CAP_PROP_FPS) or 25
W = int(cap.get(3))
H = int(cap.get(4))

out = cv2.VideoWriter(
    OUTPUT_VIDEO,
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (W, H)
)

interval_f = int(INTERVAL_SEC * fps)
on_f = int(ON_SEC * fps)
f = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    slot = (f // interval_f) % len(MESSAGE)
    if f % interval_f < on_f:
        digit = MESSAGE[slot]
        sym = DIGIT_TO_SYMBOL[digit]
        block = draw_symbol(sym)

        cy, cx = get_pos(H, W, slot)
        y, x = cy - ROI//2, cx - ROI//2
        mask = block[:,:,0] > 0
        frame[y:y+ROI, x:x+ROI][mask] = block[mask]

    out.write(frame)
    f += 1

cap.release()
out.release()

print("Encoded message:", MESSAGE)
```

```python
import cv2
import numpy as np
from collections import Counter

VIDEO = "/content/recorded_encoded_123456.mp4"

# CHANGE ONLY THIS EXPECTED MESSAGE
EXPECTED_MESSAGE = [1,2,3,4,5,6]

DIGIT_TO_SYMBOL = {
    1: "+",
    2: "-",
    3: "x",
    4: "|",
    5: "--",
    6: "||"
}

EXPECTED_OPS = set(DIGIT_TO_SYMBOL[d] for d in EXPECTED_MESSAGE)

LOW_BLUE  = np.array([90, 80, 80])
HIGH_BLUE = np.array([130, 255, 255])

def classify_operator(roi):
    gray = cv2.cvtColor(roi, cv2.COLOR_BGR2GRAY)
    edges = cv2.Canny(gray, 50, 150)

    lines = cv2.HoughLines(edges, 1, np.pi/180, 80)
    if lines is None:
        return None

    angles = [l[0][1] for l in lines]
    vert  = sum(abs(a - np.pi/2) < 0.12 for a in angles)
    horiz = sum(abs(a) < 0.12 for a in angles)
    diag  = sum(abs(a - np.pi/4) < 0.15 or abs(a - 3*np.pi/4) < 0.15 for a in angles)

    if vert >= 2:
        return "||"
    if horiz >= 2:
        return "--"
    if vert and horiz:
        return "+"
    if diag >= 2:
        return "x"
    if vert:
        return "|"
    if horiz:
        return "-"
    return None

cap = cv2.VideoCapture(VIDEO)
detected_ops = []

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv, LOW_BLUE, HIGH_BLUE)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, np.ones((5,5),np.uint8))

    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    for cnt in contours:
        x,y,w,h = cv2.boundingRect(cnt)
        if w*h < 2500:
            continue

        roi = frame[y:y+h, x:x+w]
        op = classify_operator(roi)
        if op:
            detected_ops.append(op)

cap.release()

counts = Counter(detected_ops)
print("Detected operator counts:", counts)

missing = EXPECTED_OPS - set(counts.keys())

if not missing:
    print("✅ WATERMARK VERIFIED")
    print("Matched payload:", "".join(map(str, EXPECTED_MESSAGE)))
else:
    print("❌ WATERMARK NOT VERIFIED")
    print("Missing operators:", missing)
```
